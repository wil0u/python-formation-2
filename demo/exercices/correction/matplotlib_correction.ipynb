# Exercice 3
# Reprenez le graphique précédent et ajoutez un titre, des étiquettes pour les axes (étiquette x : "Nombres", étiquette x : "Carrés"),
# changez la couleur de la ligne en rouge et ajoutez un marqueur pour les points en forme d'étoile.

# Exercice 4
# Superposez deux graphiques : Le graphique précédant et celui réaliser à l'exercice 1   
# Ajouter un titre (: "Superposisition de deux graphiques"), un label à l'axe des abscisse ('Nombres'), un label à l'axe des ordonnées ('y') 


```python

import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 11)

# Premier graphique : y = x^2
y1 = x ** 2
plt.plot(x, y1, color='red', marker='*', label='y = x^2')  # Change color to red and add marker

# Deuxième graphique : y = x
y2 = x
plt.plot(x, y2, label='y = x')

plt.title('Superposition de deux graphiques')  # Add title
plt.xlabel('Nombres')  # Add x-axis label
plt.ylabel('y')  # Add y-axis label
plt.legend(loc='upper left')  # Add legend
plt.show()
```


# Exercice 6
# Réaliser le graphique se trouvant sur la figure ci-dessous, (l'annotations des coordonnées de chaque point est optionnelle) 


```python
x = [1,3,3,5,5,7]
y = [1,1,3,3,5,5]
plt.plot(x, y, color='green')  # Make a step plot with green lines using plt.plot
plt.title('Graphique en escalier')  # Add title

# Annotate points with their coordinates using plt.annotate
for i, j in zip(x, y):
    plt.annotate(f'({i},{j})', (i, j))

#plt.savefig('escalier.png')
plt.show()
```

# Exercice 7
# Créez un graphique de dispersion (scatter plot) pour montrer la relation entre la longueur et la largeur des pétales 
# dans le dataset Iris.

```python

from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import pandas as pd

# Charger le dataset Iris
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Créer le scatter plot
plt.scatter(iris_df['petal length (cm)'], iris_df['petal width (cm)'])
plt.title('Graphique de dispersion de la longueur vs largeur des pétales')
plt.xlabel('Longueur des pétales (cm)')
plt.ylabel('Largeur des pétales (cm)')
plt.show()
```


# Exercice 8
# Reprennez les graphiques de la question 4, mais cette fois alignez les graphiques horizontalement l'un à côté de l'autre. 
# Ajoutez un titre à la figure ("Deux graphiques alignés horizontalement"), un label à l'axe des abscisses ('Nombres') et un label à l'axe des ordonnées ('y').

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 11)

# Création de la figure et des subplots
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10, 5))
fig.suptitle('Superposition de deux graphiques')

# Premier graphique : y = x^2
y1 = x ** 2
axs[0].plot(x, y1, color='red', marker='*')  # Change color to red and add marker
axs[0].set_title('y = x^2')
axs[0].set_xlabel('Nombres')
axs[0].set_ylabel('y')

# Deuxième graphique : y = x
y2 = x
axs[1].plot(x, y2)
axs[1].set_title('y = x')
axs[1].set_xlabel('Nombres')
axs[1].set_ylabel('y')

plt.tight_layout()
plt.show()
```
