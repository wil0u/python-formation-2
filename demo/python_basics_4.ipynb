Utilisation de abs()<br>
abs() retourne la valeur absolue d'un nombre

```python
num = -10
print(abs(num))  # 10
```

Utilisation de all()<br>
all() retourne True si tous les éléments dans un itérable sont vrais, sinon il retourne False

```python
liste = [1, True, True, True]
print(all(liste))  # True
```

Utilisation de any()<br>
any() retourne True si au moins un élément d'un itérable est vrai, sinon il retourne False

```python
liste = [0, False, True]
print(any(liste))  # True
```

Utilisation de bin()<br>
bin() convertit un nombre entier en une chaîne de caractères binaire

```python
num = 5
print(bin(num))  # '0b101'
```

Utilisation de chr()<br>
chr() retourne une chaîne représentant un caractère à un point de code Unicode

```python
code = 97
print(chr(code))  # 'a'
```

Utilisation de dir()<br>
dir() tente de retourner une liste d'attributs valides pour cet objet

```python
print(dir(num))
```

Utilisation de enumerate()<br>
enumerate() retourne un objet énumérable

```python
for i, val in enumerate(['a', 'b', 'c']):
    print(i, val)  # 0 a, 1 b, 2 c
```

Utilisation de isinstance()<br>
isinstance() vérifie si un objet est une instance d'une classe

```python
print(isinstance(5, int))  # True
```

Utilisation de len()<br>
len() retourne la longueur (le nombre d'éléments) d'un objet

```python
print(len('Python'))  # 6
```

Utilisation de max()<br>
max() retourne le plus grand élément

```python
print(max([1, 2, 3]))  # 3
```

Utilisation de min()<br>
min() retourne le plus petit élément

```python
print(min([1, 2, 3]))  # 1
```

Utilisation de oct()<br>
oct() convertit un nombre entier en une chaîne octale

```python
print(oct(8))  # '0o10'
```

Utilisation de hex()<br>
hex() convertit un nombre entier en une chaîne hexadécimale

```python
print(hex(255))  # '0xff'
```

Utilisation de print()<br>
print() affiche l'argument à l'écran

```python
print('Hello, world!')  # Hello, world!
```

Utilisation de range()<br>
range() génère une séquence de nombres

```python
for i in range(5):
    print(i)  # 0, 1, 2, 3, 4
```

Utilisation de round()<br>
round() arrondit un nombre

```python
print(round(5.76543, 2))  # 5.77
```

Utilisation de open()<br>
open() ouvre un fichier et retourne un objet fichier<br>
Ici on ouvre un fichier en mode écriture et lecture ('w+')<br>
De plus si le fichier n'existe pas alors il est créer, et si il existe déjà alors son contenu est écrasé.

```python
with open('file.txt', 'w+') as file:
    file.writelines("Hello World !")
    #Permet de replacer le curseur au début du fichier.
    file.seek(0)
    print(file.readlines())
```

Utilisation de type()<br>
type() retourne le type d'un objet

```python
print(type(5))  # <class 'int'>
```

Utilisation de sorted()<br>
sorted() retourne une nouvelle liste triée

```python
print(sorted([3, 1, 2]))  # [1, 2, 3]
```

Utilisation de sum()<br>
sum() additionne tous les éléments d'un itérable

```python
print(sum([1, 2, 3]))  # 6
```
