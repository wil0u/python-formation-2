###### fonction qui cherche un mot

```python
# attention il faut chercher le mot entier
# et choisir comment gérer la casse et les accents

# instruction in la plus rapide
def contains_word(chaine, word):
    return (' ' + word.upper() + ' ') in (' ' + chaine.upper() + ' ')
```

```python
# 16 Go ou 16go ou 1to

import re
def test(w):
    if re.search("[0-9]{1,3}\s{0,1}[gt][ob]",w.lower()):
        return re.findall("[0-9]{1,3}\s{0,1}[gt][ob]",w.lower())
```

```python
# fonction qui teste si une chaine contient 2 mots contenant uniquement des lettres et de longueur 1 à 3
def contains2AlphaMots13(s):
    res = 0
    if len(re.findall(r"\b[a-zA-Z]{1,3}\b",s)) >= 2:
        res =  1
    return res
```

```python
contains2AlphaMots13("xxxxxxxxxxxxxx MOT 12 zeeze")
```

```python
contains2AlphaMots13("zx12 MOT AAA 425")
```

```python
contains2AlphaMots13("zx MOT AAAAAA 425")
```

```python
# fonction qui checke si un mot ressemble à un code postal fr
def isCodePostal(word):
    res = False
    if word.strip().isdigit():
        if len(word.strip())==5:
            res = True
    return res
```

```python
isCodePostal("92000")
```

```python
isCodePostal("92000000")
```

## re.compile

```python
# vérifier qu'un mot de passe fait au moins 8 caractères 
# et qu'il ne contient que des lettres majuscules, minuscules et des chiffres
chn_mdp = r"^[A-Za-z0-9]{8,}$"
exp_mdp = re.compile(chn_mdp)
mot_de_passe = ""
while exp_mdp.search(mot_de_passe) is None :
    mot_de_passe = input("Tapez votre mot de passe : ")
```
