# Les éléments qui composent une figure matplotlib

# Un simple nuage de points

```python
plt.scatter([1,2,3,4],[1,4,9,16],marker='*')
plt.show()
```

# Un graphique superposant plusieurs droites 

```python
plt.plot([ 1, 2, 3, 4], [1, 4, 9, 16],'bo')
plt.plot([ 1, 2, 3, 4], [2,6, 12, 18],marker='o',color='green', linestyle='dashed',markersize=12)
plt.plot([ 1, 2, 3, 4], [2**2,6**2, 12**2, 18**2],marker='o',color='green', linestyle='dashed',markersize=12)
plt.show()
```

# Un simple bar chart

```python
plt.bar([1,2,3,4],[1,4,9,16], tick_label=["label 1","label 2","label 3","label 4"], width=-0.8, align='edge', edgecolor='red', linewidth=4.5)
plt.show()
```

# Représenter une loi normale avec la fonction hist

```python
mu = 0
sigma = 1
normal_law = np.random.normal(mu,sigma,size=10000)
_,bins,_=  plt.hist(normal_law,density=True,bins=100)
y = ((1 / (np.sqrt(2 * np.pi) * sigma)) *
     np.exp(-0.5 * (1 / sigma * (bins - mu))**2))
plt.plot(bins, y, '--')
plt.show()
```

# Un graphique plus complexe et customisé

```python

x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
z = [2*value**2 for value in x]
# Définir la taille et la résolution du graphique
plt.figure(figsize=(15,11), dpi=180)

# Définir les titres du graphique et des axes
plt.title("Mon Graphique Complexifié")
plt.xlabel("Axe des x")
plt.ylabel("Axe des y")

# Définir les limites des axes
plt.xlim(0, 5)
plt.ylim(0, 35)

# Ajouter une grille
plt.grid(True)

# Changer la couleur, le type de ligne, l'épaisseur de ligne et le marqueur
plt.plot(x, y, color="red", linestyle="--", linewidth=2, marker=">", markersize=8, label="y = x**2")
plt.plot(x, z, color="green", linestyle=":", linewidth=2, marker="*", markersize=12, label="z = 2x**2")
# Ajouter une légende
plt.legend()
annotation_list = ["Paris","New-York","Madrid","Berlin"]
# Ajouter des annotations
for i in range(len(x)):
    plt.annotate(f'({annotation_list[i]})', (x[i], y[i]), ha='center')

# Afficher le graphique
plt.show()
```

# La différence entre créer des graphes de manière implicite et explicite


#
# Coding styles
# =============
#
# The explicit and the implicit interfaces
# ----------------------------------------
#
# As noted above, there are essentially two ways to use Matplotlib:
#
# - Explicitly create Figures and Axes, and call methods on them (the
#   "object-oriented (OO) style").
# - Rely on pyplot to implicitly create and manage the Figures and Axes, and
#   use pyplot functions for plotting.
#
# See :ref:`api_interfaces` for an explanation of the tradeoffs between the
# implicit and explicit interfaces.
#
# So one can use the OO-style


```python
import numpy as np
x = np.linspace(0, 2, 100)  # Sample data.

plt.figure(figsize=(5, 2.7))
plt.plot(x, x, label='linear')  # Plot some data on the (implicit) Axes.
plt.plot(x, x**2, label='quadratic')  # etc.
plt.plot(x, x**3, label='cubic')
plt.xlabel('x label')
plt.ylabel('y label')
plt.title("Simple Plot")
plt.legend()
plt.show() 

```

```python

x = np.linspace(0, 2, 100)  # Sample data.
# Note that even in the OO-style, we use `.pyplot.figure` to create the Figure.
fig, ax = plt.subplots(figsize=(5, 2.7))
ax.plot(x, x, label='linear')  # Plot some data on the Axes.
ax.plot(x, x**2, label='quadratic')  # Plot more data on the Axes...
ax.plot(x, x**3, label='cubic')  # ... and some more.
ax.set_xlabel('x label')  # Add an x-label to the Axes.
ax.set_ylabel('y label')  # Add a y-label to the Axes.
ax.set_title("Simple Plot")
ax.legend()# Add a legend.

# Le mode explicit permet par exemple de customiser les major ticks et les spines

#ax.xaxis.set_major_locator(plt.MultipleLocator(0.5))
#ax.yaxis.set_major_locator(plt.MultipleLocator(2))
#ax.spines['bottom'].set_position('center')
# Add a title to the Axes.

plt.show()
#fig.savefig('myfig')
```

```python
fig, ax = plt.subplots(figsize=(25, 15),ncols=3,nrows=2)
ax[0,0].scatter([1,2,3,4],[1,4,9,16])

ax[0,1].plot([ 1, 2, 3, 4], [1, 4, 9, 16],'bo')
ax[0,1].plot([ 1, 2, 3, 4], [2,6, 12, 18],marker='o',color='green', linestyle='dashed',markersize=12)
ax[0,1].plot([ 1, 2, 3, 4], [2**2,6**2, 12**2, 18**2],marker='o',color='green', linestyle='dashed',markersize=12)

ax[0,2].bar([1,2,3,4],[1,4,9,16], tick_label=["hello","test","test2","test3"],width=-0.8,align='edge',edgecolor='red',linewidth=4.5)

mu = 0
sigma = 1
normal_law = np.random.normal(mu,sigma,size=10000)
_,bins,_=  ax[1,0].hist(normal_law,density=True,bins=10)
y = ((1 / (np.sqrt(2 * np.pi) * sigma)) *
     np.exp(-0.5 * (1 / sigma * (bins - mu))**2))
ax[1,0].plot(bins, y, '--')


x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
z = [p*2 for p in y]
# Définir la taille et la résolution du graphique

# Définir les titres du graphique et des axes
ax[1,1].set_title("Mon Graphique Complexifié")
ax[1,1].set_xlabel("Axe des x")
ax[1,1].set_ylabel("Axe des y")

# Définir les limites des axes
ax[1,1].set_xlim(0, 5)
ax[1,1].set_ylim(0, 35)

# Ajouter une grille
ax[1,1].grid(True)

# Changer la couleur, le type de ligne, l'épaisseur de ligne et le marqueur
ax[1,1].plot(x, y, color="red", linestyle="--", linewidth=2, marker="o", markersize=8, label="y = x**2")
ax[1,1].plot(x, z, color="green", linestyle="-", linewidth=2, marker="*", markersize=12, label="y = y*2")
# Ajouter une légende
ax[1,1].legend()
annotation_list = ["Test1","Test2","Test3","Test4"]
# Ajouter des annotations
for i in range(len(x)):
    ax[1,1].annotate('({})'.format(annotation_list[i]), (x[i], y[i]), textcoords="offset points", xytext=(-10,-10), ha='center')


x = np.linspace(0, 2, 100)  # Sample data.    
ax[1,2].plot(x, x, label='linear')  # Plot some data on the Axes.
ax[1,2].plot(x, x**2, label='quadratic')  # Plot more data on the Axes...
ax[1,2].plot(x, x**3, label='cubic')  # ... and some more.
ax[1,2].set_xlabel('x label')  # Add an x-label to the Axes.
ax[1,2].set_ylabel('y label')  # Add a y-label to the Axes.
ax[1,2].set_title("Simple Plot")  # Add a title to the Axes.
ax[1,2].legend()# Add a legend.

fig.savefig("composition_de_graphe")
```

# Exemple avec une dataframe et un jeu de données réel

```python
from sklearn.datasets import load_iris
import pandas as pd

# Load the Iris dataset
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df
```

# Graphe qui affiche la proportion des espèces d'iris présent dans le jeu de donnée IRIS.

```python
# Add a column for the species - convert the species from integers to the actual species names
iris_df['species'] = [iris.target_names[i] for i in iris.target]

# Count the number of each species
species_counts = iris_df['species'].value_counts()

# Create the pie chart
plt.figure(figsize=(10, 6))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle
plt.title('Distribution des espèces d\Iris.')
plt.show()
```

# Nuage de points : de la longueur du sepal vs la largeur du sepal.

```python
# Let's make a scatter plot of sepal length vs sepal width
plt.figure(figsize=(10, 6))
plt.scatter(iris_df['sepal length (cm)'], iris_df['sepal width (cm)'])
plt.title('Nuage de point : longueur de sépal (sepal length) vs largeur de sépal (sepal width)')
plt.xlabel('Sepal length (cm)')
plt.ylabel('Sepal width (cm)')
plt.show()
```
