```python
# Introduire une gestion d'exception dans la fonction => try except
def standardisation(x): 
    try:        
        x-=np.mean(x) # the -= means can be read as x = x- np.mean(x)
        x/=np.std(x)
        return x
    except TypeError:
        print("le type des données n'est pas float")
```

```python
standardisation(Matrice)
```

```python
# si matrice2 n'est pas un array mais un tuple
# cela marche car np.mean fonctionne sur une liste de val numériques
matrice2=(1,2,3,4)
print(type(matrice2))
standardisation(matrice2)
print(np.mean(matrice2))
```

```python
# en classe
import numpy as np
# # constructeur de la classe : ce qui est fait quand on appelle la classe
class maClasseStandardisation():
    # self fait ref a une instance dobjet, x est linput
    def __init__(self, x):
        self.x = x
        
        # on declare lattribut standardisation
        self.xnorm = x
        self.err=-1
    
    # fonction principale, standardisation
    def standardisation(self):
        try:            
            self.xnorm=self.xnorm-np.mean(self.x)
            self.xnorm=self.xnorm/np.std(self.x)
            return self #quand on appell on retourne tout
        except TypeError:
            print("le type des données n'est pas float")
            self.err=999 # on def un code err
            return self # on retourne les resultats quand lexcept est raise pour pouvoir afficher le print (__str__)
        
    # controle le print
    def __str__(self):
        return f"origine : {self.x}, normée: {self.xnorm}, Code Erreur: {self.err}"
```

```python
matrice=(1.1,2.2,3.5,4.8)
print(maClasseStandardisation(matrice))
```

```python
print(maClasseStandardisation(matrice).standardisation())
```

## Lambda Function

```python
# appliquons une lambda fonction pour mettre au carré les valeurs du vecteur numérique
# utilise la lambda func pour à tout x renvoyer f(x)=x²
# map applique une fonction à un iterable, ici le tuple numbers
map(lambda x :x**2,Matrice)
```

```python
# fonction qui test esi un element vaut 1
result= list(map(lambda x: x==1.00, Matrice))
print(result)
result.count(False)
```

```python
#python ne sait pas représenter l'instance donc il faut la transformer
list(map(lambda x :x**2,Matrice))
```

```python
# Créer une fonction qui retourne l'aire d'un triangle et l'appliquer pour un triangle de base 2 et de hauteur 3
f = lambda base, hauteur : (base*hauteur)/2
f(base = 2,hauteur = 3)
```
