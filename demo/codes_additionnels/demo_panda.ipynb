```python
import pandas as pd
import numpy as np
```

----------------------------------------------------------
1. Chargement des données 
----------------------------------------------------------

Création de données d'exemple

```python
data1 = {
    'ID': [1, 2, 3, 4],
    'Nom': ['Alice', 'Bob', 'Charlie', 'David'],
    'Departement': ['A', 'B', 'A', 'C'],
    'Salaire': [50000, 55000, None, 60000]
}
```

```python
data1
```

```python
data2 = {
    'ID': [1, 2, 3, 4],
    'Projets': [3, 5, 2, 4],
    'Competences': ['Python', 'SQL', 'Java', 'Python']
}
```

Instanciation des dataframes à partir d'un dictionnaire

```python
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
```

```python
print("--- DataFrame 1 ---")
print(df1)
print("\n--- DataFrame 2 ---")
print(df2)
```

----------------------------------------------------------
2. Fusion de dataframes
----------------------------------------------------------

Fusionner les données sur la clé commune ID

```python
df_merged = pd.merge(df1, df2, on='ID', how='inner')
print("DataFrame fusionné :")
print(df_merged)
```

----------------------------------------------------------
3. Néttoyage des données
----------------------------------------------------------

Renommer une colonne

```python
df_merged.rename(columns={'Competences': 'Langage'}, inplace=True)
```

Gérer les valeurs manquantes

```python
df_merged['Salaire'] = df_merged['Salaire'].fillna(df_merged['Salaire'].mean())
```

Convertir les types de données

```python
df_merged['ID'] = df_merged['ID'].astype('category')
```

```python
print("Après nettoyage :")
print(df_merged.dtypes)
```

----------------------------------------------------------
4. Manipulations de base
----------------------------------------------------------

Filtrer les données

```python
df_filtered = df_merged[df_merged['Projets'] >= 3]
print(df_merged)
df_filtered
```

```python
df_merged['Projets'] >= 3
```

Créer une nouvelle colonne

```python
df_filtered['Bonus'] = df_filtered['Salaire'] * 0.1
df_filtered['my_nouvelle_colonne'] = "coucou" 
```

```python
print("\nDonnées filtrées avec bonus :")
print(df_filtered)
```

----------------------------------------------------------
5. Analyse et agrégations
----------------------------------------------------------

Statistiques descriptives

```python
print("\nStatistiques salariales :")
print(df_merged.info())
```

Agrégations par département

```python
agg_results = df_merged.groupby('Langage').agg({
    'Salaire': 'mean',
    'Projets': 'sum',
    'ID': 'count'
}).rename(columns={'ID': 'Effectif'})
```

```python
print("\nAgrégations par département :")
print(agg_results)
```

Tableau croisé avec les modalités des variables "Departement" et "Competence".

```python
crosstab_result = pd.crosstab(df_merged['Departement'], df_merged['Langage'])
```

```python
print("Tableau croisé :")
crosstab_result
```

----------------------------------------------------------
6. Sauvegarde des résultats
----------------------------------------------------------

```python
df_merged.to_csv('donnees_fusionnees.csv', index=False)
print("Fichier sauvegardé avec succès !")
```
