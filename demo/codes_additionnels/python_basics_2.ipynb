-- Les mots clés réservés --

En Python, les mots clés réservés sont des noms prédéfinis qui ont un sens spécial pour l'interpréteur Python. <br>
Ce sont des mots qui ne peuvent pas être utilisés comme identifiants de variable, de fonction, de classe ou autre, <br>
car ils sont "réservés" à des utilisations spécifiques dans le langage de programmation Python. 

Importe le module math pour utiliser ses fonctions

```python
import math
print(math.sqrt(16))
```

Importe uniquement la fonction sqrt. Peut être utilisé sans le préfixe "math"   

```python
from math import sqrt
print(sqrt(16))  # Affiche 4.0
```

Import math et déclare un alias : m permet d'identifié "math"

```python
import math as m
print(m.sqrt(16))
```

Définit une fonction qui ajoute deux nombres

```python
def add(a, b):
   return a + b
```

Vérifie si un nombre est positif, négatif ou zéro

```python
num = 3.4
num = 0
if num > 0:
   print("Positive number")
elif num == 0:
   print("Zero")
else:
   print("Negative number")
```

Affiche tous les éléments d'une liste

```python
for i in [1, 2, 3, 4, 5]:
   print(i)
```

Affiche les nombres de 1 à 5

```python
i = 1
while i <= 5:
   print(i)
   i += 1
```

Utilise les opérateurs logiques and, or et not

```python
a = True
b = False
print(a and b)  # Affiche False
print(a or b)   # Affiche True
print(not a)    # Affiche False
```

Utilise les valeurs booléennes True et False

```python
a = True
if a:
   print("a est vrai")
else:
   print("a est faux")
```

Vérifie si un élément est dans une liste

```python
if 3 in [1, 2, 3, 4, 5]:
   print("3 est dans la liste")
```

Vérifie si deux variables pointent vers le même objet

```python
a = [1, 2, 3]
b = a
print(a is b)  # Affiche True
```

Utilise la valeur None pour représenter une absence de valeur

```python
a = None
if a == None:
   print("a n'a pas de valeur")
```
