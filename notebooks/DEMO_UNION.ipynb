## CONCATENATION DE TYPE UNION

```python
import pandas as pd
```

```python
# Deux Tables dont on veut faire l'UNION
df1 = pd.DataFrame([['A1','A2'],['Paris']],columns=['ID','VILLE'])
df2 = pd.DataFrame(['A3','A4'],columns=['ID'])
df3 = pd.DataFrame([['A5','A6'],['Londres']],columns=['ID2','VILLE'])
print(df1)
print(df2)
print(df3)
```

```python
print(df1.columns)
```

```python
print(df1)
print(df2)
```

```python
# UNION
pd.concat([df1, df2,df3],axis=0)
```

```python
# APPEND
df1.append(df2).append(df3)
```
