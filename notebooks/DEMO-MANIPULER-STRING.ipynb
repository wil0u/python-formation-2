```python
# soit une chaine de caractère avec des saut de lignes
text="""
ligne1 bla bla bla BLA bla BLA xxxx yy yy  BLA bla XXXX YY yy YY ; /[]
ligne2 bla BLA BLA xx XX xx 
ligne3 bla BLA BLA xx XX xx nnnnnnnnnnnnnnnnn
ligne4 - (;;,)
"""
```

```python
# supprime quelques car dont les retours chariots
cara_asupprimer='-.,\n'
for char in cara_asupprimer:
    #print(char)
    text2=text.replace(char,' ')
```

```python
text2
```

```python
# wordcount
from collections import Counter
# word count sur lower case
# 3 premiers mots
Counter(text2.lower().split(' ')).most_common(3)
```

```python
# word count sur les mots en filtrant que sur les mots en MAJUSCULES
text3=text2.split(' ')
```

```python
text_withonlyUPCASE=[word for word in text3 if word.isupper()]
Counter(text_withonlyUPCASE).most_common(3)
```

```python
# fonction qui teste si une chaine contient 2 mots contenant uniquement des lettres et de longueur 2 ou 3
# on met un espace devant et un espace à la fin pour etre sur de prendre toute la chaine

def contains2AlphaMots23(s):
    if re.search("(\s{1}[A-Z]{1,3}\s+[A-Z]{1,3}\s{1})",s):
        return 1
    else:
        return 0
```

```python
contains2AlphaMots23("xxxxxxxxxxxxxx MOT 12 zeeze")
```

```python
contains2AlphaMots23("zx12 MOT AAA 425")
```

```python
contains2AlphaMots23("zx12 MOT AAAAAA 425")
```

```python

# fonction qui checke si un mot ressemble à un code postal fr
def isCodePostal(word):
    if word.strip().isdigit():
        if len(word.strip())==5:
            return True
```

```python
isCodePostal("92000")
```

```python
isCodePostal("92000000")
```
