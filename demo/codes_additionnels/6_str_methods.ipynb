Les méthodes applicables aux chaines (STR)

On va démontrer les méthodes de string en Python qui sont utiles<br>

1. S.count(sub[, start[, end]])<br>
Description: Return the number of non-overlapping occurrences of substring sub in string S[start:end].<br>
Exemple:

```python
ma_chaine = "Hello world hello"
```

```python
ma_chaine.count("hello")
```

Avec start et end:

```python
ma_chaine.count("hello", 6, len(ma_chaine))  # Cherche a partir de l'index 6.
```

2. S.find(sub [,start [,end]])<br>
Description: Return the lowest index in S where substring sub is found, such that sub is contained within S[start:end]. Return -1 on failure.<br>
Exemple:

```python
ma_chaine.find("world")
```

Si la sous-chaine n'est pas trouvée:

```python
ma_chaine.find("absent")
```

avec start et end :

```python
ma_chaine.find("hello", 8, len(ma_chaine)) #Cherche a partir de l'index 8
```

3. S.index(sub [,start [,end]])<br>
Description: Like S.find() but raise ValueError when the substring is not found.<br>
Exemple:

```python
ma_chaine.index("world")
```

4. S.lower()<br>
Description: Return a copy of the string S converted to lowercase.<br>
Exemple:

```python
ma_chaine.lower()
```

5. S.replace(old, new[, count])<br>
Description: Return a copy of string S with all occurrences of substring old replaced by new. If the optional argument count is given, only the first count occurrences are replaced.<br>
Exemple:

```python
ma_chaine.replace("hello", "bonjour")
```

Avec count:

```python
ma_chaine.replace("hello", "bonjour", 1) #Seulement le premier hello est remplacÃ©
```

6. S.split([sep [,maxsplit]])<br>
Description: Return a list of the words in the string S, using sep as the delimiter string. If maxsplit is given, at most maxsplit splits are done. If sep is not specified or is None, any whitespace string is a separator and empty strings are removed from the result.<br>
Exemple:

```python
ma_chaine.split() #Sep par defaut est l'espace
```

Avec un separateur spécifique:

```python
ma_chaine_sep = "a,b,c,d"
ma_chaine_sep.split(",")
```

7. S.strip([chars])<br>
Description: Return a copy of the string S with leading and trailing whitespace removed.<br>
Exemple:

```python
chaine_avec_espaces = "   Hello world   "
chaine_avec_espaces.strip()
```

Avec des caractères à supprimer:

```python
chaine_speciale = "***Hello world***"
chaine_speciale.lstrip("*")
```

8. S.upper()<br>
Description: Return a copy of the string S converted to uppercase.<br>
Exemple:

```python
ma_chaine.upper()
```
