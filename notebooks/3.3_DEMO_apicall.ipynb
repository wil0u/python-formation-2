## COMPLEMENT JSON et DICTIONNAIRE
Utile pour sauvegarder des listes de paramètres, des mots de passe
Utilisé en numérique pour envoyer et recevoir des données 

```python
import slumber
Adresse="http://api-adresse.data.gouv.fr"
API=slumber.API(Adresse)
resultat=API.search.get(q="55 quai de grenelle, PARIS")
```

```python
resultat
```

```python
resultat.keys()
```

```python
resultat["features"]
# l'objet est une liste
```

```python
resultat["features"][0]['geometry']
```

```python
# on cree un object dict qui contient deux clés
d = {"coord1":resultat["features"][0]['geometry']['coordinates'][0], "coord2":resultat["features"][0]['geometry']['coordinates'][1] }
print(d)
```

```python
# on veut le sauvegarder en json
import json
# on sauvegarde le dictionnaire avec la librairie json
with open("exemple.json", "w") as f:
    json.dump(d, f)
```
