## Démo sur les "function"s en python

#### Paramètres positionnels versus Paramètres nommés

```python
# Arguments positionnels (l'ordre compte)
def decrire_animal(nom, age, espece):
    """Fonction avec arguments positionnels"""
    print(f"{nom} est un {espece} de {age} ans")
```

```python
decrire_animal("Médor", 5, "chien")
decrire_animal("chat", "Félix", 3)  # Problème: l'ordre est incorrect
```

```python
# Arguments nommés (l'ordre ne compte pas)
decrire_animal(age=2, espece="poisson", nom="Bubulle")
decrire_animal(nom="Minou", age=4, espece="chat")

# Combinaison d'arguments positionnels et nommés
decrire_animal("Rex", age=7, espece="chien")
```

```python
# Paramètres avec valeurs par défaut
def decrire_animal_amelioré(nom, espece, age=1):
    """Fonction avec paramètre par défaut"""
    print(f"{nom} est un {espece} de {age} ans")
```

```python
decrire_animal_amelioré("Titi", "canari")  # Utilise la valeur par défaut pour age
decrire_animal_amelioré("Hector", "tortue", 10)  # Spécifie une valeur pour age
```

#### L'utilisation de "*args" et "**kwargs"

```python
# *args pour accepter un nombre variable d'arguments positionnels
def additionner(*args):
    """Additionne un nombre variable de valeurs"""
    print(f"Arguments reçus: {args}")
    return sum(args)
```

```python
print(additionner(1, 2, 3))
print(additionner(10, 20, 30, 40, 50))
```

```python
# **kwargs pour accepter un nombre variable d'arguments nommés
def decrire_personne(**kwargs):
    """Affiche les informations d'une personne"""
    print("Informations sur la personne:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
decrire_personne(nom="Alice", age=25, ville="Paris", sexe="Féminin")
```

```python
decrire_personne(nom="Bob", profession="Ingénieur", pays="France")
```

```python
# Combinaison d'arguments normaux, *args et **kwargs
def fonction_complete(a, b, *args, option=True, **kwargs):
    """Fonction utilisant tous les types d'arguments"""
    print(f"a: {a}, b: {b}")
    print(f"args: {args}")
    print(f"option: {option}")
    print(f"kwargs: {kwargs}")
    print("---")
```

```python
fonction_complete(1, 2)
fonction_complete(1, 2, 3, 4, 5)
fonction_complete(1, 2, 3, 4, 5, option=False)
fonction_complete(1, 2, 3, 4, 5, option=True, nom="Alice", age=30)
```

### La portée des variables 

```python
# Variable globale
variable_globale = "Je suis globale"
```

```python
def test_portee():
    """Test de la portée des variables"""
    # Variable locale (différente de la globale)
    variable_locale = "Je suis locale"
    print(f"À l'intérieur de la fonction: {variable_locale}")
    
    # Accéder à une variable globale
    print(f"Accès à la variable globale: {variable_globale}")
    
    # Modifier une variable globale (nécessite le mot-clé global)
    #variable_globale = "J'ai été modifiée"
    print(f"Variable globale modifiée: {variable_globale}")
```

```python
print(f"Avant la fonction: {variable_globale}")
test_portee()
print(f"Après la fonction: {variable_globale}")
```

```python
# La variable locale n'est pas accessible en dehors de la fonction
try:
    print(variable_locale)
except NameError as e:
    print(f"Erreur: {e}")
```

```python
# Exemple plus complexe de portée
variable_externe = "externe"
```

```python
def portee_imbriquee():
    """Démonstration de portée imbriquée"""
    variable_interne = "interne"
    
    def fonction_interne():
        """Fonction à l'intérieur d'une autre fonction"""
        variable_plus_interne = "plus interne"
        print(f"Fonction interne: {variable_externe}")
        print(f"Fonction interne: {variable_interne}")
        print(f"Fonction interne: {variable_plus_interne}")
        
    
    fonction_interne()
    print(f"Fonction externe: {variable_externe}")
    print(f"Fonction externe: {variable_interne}")
    
    # La variable de la fonction interne n'est pas accessible ici
    try:
        print(variable_plus_interne)
    except NameError as e:
        print(f"Erreur: {e}")
```

```python
portee_imbriquee()
```
