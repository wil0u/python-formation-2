```python
# en classe
import numpy as np
# # constructeur de la classe : ce qui est fait quand on appelle la classe
class maClasseStandardisation():
    # self fait ref a une instance dobjet, x est linput
    def __init__(self, x):
        self.x = x
        
        # on declare lattribut standardisation
        self.xnorm = x
        self.err=-1
    
    # fonction principale, standardisation
    def standardisation(self):
        try:            
            self.xnorm=self.xnorm-np.mean(self.x)
            self.xnorm=self.xnorm/np.std(self.x)
            return self #quand on appell on retourne tout
        except TypeError:
            print("le type des données n'est pas float")
            self.err=999 # on def un code err
            return self # on retourne les resultats quand lexcept est raise pour pouvoir afficher le print (__str__)
        
    # controle le print
    def __str__(self):
        return f"origine : {self.x}, normée: {self.xnorm}, Code Erreur: {self.err}"
```

```python
matrice=(1.1,2.2,3.5,4.8)
mon_objet = maClasseStandardisation(matrice)
```
