## CONCATENATION DE TYPE UNION

```python

import pandas as pd
```

```python
# Deux Tables dont on veut faire l'UNION
df1 = pd.DataFrame([['A1','A2'],['Paris']])
df2 = pd.DataFrame(['A3','A4'])
df3 = pd.DataFrame([['A5','A6'],['Londres']])
print(df1)
print(df2)
```

```python
# Les col s'appellent pas pareil
df1.columns=['ID']
df2.columns=['ID2']
df3.columns=['ID']
```

```python
print(df1)
print(df2)
```

```python
# UNION
pd.concat([df1, df2,df3],axis=0)
```

```python
# APPEND
df1.append(df2)
```
