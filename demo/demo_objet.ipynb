```python
class CompteBancaire:
    """Classe représentant un compte bancaire avec plafonds de dépenses"""
    
    def __init__(self, titulaire, solde_initial=0, plafond_retrait=1000, plafond_journalier=500):
        # Attributs d'instance
        self.titulaire = titulaire
        self.solde = solde_initial
        self.plafond_retrait = plafond_retrait  # Plafond par opération
        self.plafond_journalier = plafond_journalier  # Plafond total par jour
        self.retraits_aujourdhui = 0
        self.historique = [f"Création du compte avec {solde_initial}€"]
        self.derniere_date = None  # Pour suivre les changements de jour
    
    def _verifier_date(self):
        """Vérifie si on a changé de jour et réinitialise les compteurs si nécessaire"""
        from datetime import date
        aujourdhui = date.today()
        
        if self.derniere_date != aujourdhui:
            self.retraits_aujourdhui = 0
            self.derniere_date = aujourdhui
            self.historique.append(f"Nouveau jour: {aujourdhui}")
    
    def deposer(self, montant):
        """Déposer de l'argent sur le compte"""
        if montant > 0:
            self.solde += montant
            self.historique.append(f"Dépôt de {montant}€")
            return f"{montant}€ déposés. Nouveau solde: {self.solde}€"
        return "Le montant doit être positif"
    
    def retirer(self, montant):
        """Retirer de l'argent du compte avec vérification des plafonds"""
        self._verifier_date()
        
        if montant <= 0:
            return "Le montant doit être positif"
        
        if self.solde < montant:
            return "Fonds insuffisants"
        
        # Vérification du plafond par opération
        if montant > self.plafond_retrait:
            return f"Plafond de retrait par opération dépassé ({self.plafond_retrait}€)"
        
        # Vérification du plafond journalier
        if self.retraits_aujourdhui + montant > self.plafond_journalier:
            return f"Plafond journalier de retrait dépassé ({self.plafond_journalier}€)"
        
        # Si toutes les vérifications sont passées
        self.solde -= montant
        self.retraits_aujourdhui += montant
        self.historique.append(f"Retrait de {montant}€")
        return f"{montant}€ retirés. Nouveau solde: {self.solde}€. Retraits aujourd'hui: {self.retraits_aujourdhui}€"
    
    def modifier_plafonds(self, nouveau_plafond_retrait=None, nouveau_plafond_journalier=None):
        """Modifier les plafonds de retrait"""
        if nouveau_plafond_retrait is not None:
            self.plafond_retrait = nouveau_plafond_retrait
            self.historique.append(f"Plafond par opération modifié à {nouveau_plafond_retrait}€")
        
        if nouveau_plafond_journalier is not None:
            self.plafond_journalier = nouveau_plafond_journalier
            self.historique.append(f"Plafond journalier modifié à {nouveau_plafond_journalier}€")
        
        return "Plafonds mis à jour"
    
    def afficher_solde(self):
        """Afficher le solde actuel"""
        return f"Solde du compte de {self.titulaire}: {self.solde}€"
    
    def afficher_plafonds(self):
        """Afficher les plafonds actuels"""
        return f"Plafond par opération: {self.plafond_retrait}€ | Plafond journalier: {self.plafond_journalier}€"
    
    def afficher_historique(self):
        """Afficher l'historique des transactions"""
        return "\n".join(self.historique)
    
    # Représentation de l'objet
    def __str__(self):
        return f"Compte de {self.titulaire} - Solde: {self.solde}€"
    
    def __repr__(self):
        return f"CompteBancaire('{self.titulaire}', {self.solde}, {self.plafond_retrait}, {self.plafond_journalier})"

# Utilisation de la classe
compte = CompteBancaire("Alice", 2000, plafond_retrait=800, plafond_journalier=1000)

print(compte)
print(compte.afficher_plafonds())

print(compte.deposer(500))
print(compte.retirer(300))
print(compte.retirer(900))  # Devrait échouer (dépasse le plafond par opération)
print(compte.retirer(800))  # Devrait réussir
print(compte.retirer(300))  # Devrait échouer (dépasse le plafond journalier)

print(compte.modifier_plafonds(nouveau_plafond_retrait=500, nouveau_plafond_journalier=1500))
print(compte.retirer(500))  # Devrait réussir avec les nouveaux plafonds

print("\nHistorique du compte:")
print(compte.afficher_historique())
```

```python
class Date:
    """Classe représentant une date avec des méthodes spéciales"""
    
    def __init__(self, jour, mois, annee):
        self.jour = jour
        self.mois = mois
        self.annee = annee
    
    # Méthode pour l'affichage
    def __str__(self):
        return f"{self.jour:02d}/{self.mois:02d}/{self.annee}"
    
    def __repr__(self):
        return f"Date({self.jour}, {self.mois}, {self.annee})"
    
    # Méthodes de comparaison
    def __eq__(self, other):
        """Égalité: date1 == date2"""
        return (self.jour, self.mois, self.annee) == (other.jour, other.mois, other.annee)
    
    def __lt__(self, other):
        """Comparaison: date1 < date2"""
        return (self.annee, self.mois, self.jour) < (other.annee, other.mois, other.jour)
    
    def __le__(self, other):
        """Comparaison: date1 <= date2"""
        return (self.annee, self.mois, self.jour) <= (other.annee, other.mois, other.jour)
    
    # Méthodes arithmétiques       
    def __add__(self, jours):
        """Ajouter des jours à une date"""
        # Version simplifiée (ne gère pas les mois de longueur variable)
        nouveau_jour = self.jour + jours
        nouveau_mois = self.mois
        nouvelle_annee = self.annee
        
        while nouveau_jour > 30:  # Approximation
            nouveau_jour -= 30
            nouveau_mois += 1
            if nouveau_mois > 12:
                nouveau_mois = 1
                nouvelle_annee += 1
        
        return Date(nouveau_jour, nouveau_mois, nouvelle_annee)
    
    def __sub__(self, other):
        """Soustraire deux dates (donne la différence en jours)"""
        # Version simplifiée
        jours_self = self.annee * 360 + self.mois * 30 + self.jour
        jours_other = other.annee * 360 + other.mois * 30 + other.jour
        return abs(jours_self - jours_other)
    
    # Méthode pour le hachage (utile pour les sets et dictionnaires)
    def __hash__(self):
        return hash((self.jour, self.mois, self.annee))
    
    # Méthode pour vérifier si la date est valide
    def est_valide(self):
        """Vérifie si la date est valide (version simplifiée)"""
        return (1 <= self.jour <= 31 and 
                1 <= self.mois <= 12 and 
                self.annee > 0)

```

```python
# Testons nos méthodes spéciales
date1 = Date(15, 5, 2023)
date2 = Date(20, 5, 2023)
date3 = Date(15, 5, 2023)
```

```python
print("Date1:", date1)
print("Date2:", date2)
print("Date3:", date3)
```

```python
print("Date1 == Date2?", date1 == date2)
print("Date1 == Date3?", date1 == date3)
print("Date1 < Date2?", date1 < date2)
print("Date1 <= Date2?", date1 <= date2)

# Addition de jours
date_future = date1 + 20
print("Date1 + 20 jours:", date_future)

# Différence entre dates
difference = date2 - date1
print("Différence entre Date2 et Date1:", difference, "jours (approximative)")
```

```python


# Utilisation dans un set (nécessite __hash__)
dates_set = {date1, date2, date3}
print("Set de dates:", dates_set)

# Vérification de validité
date_invalide = Date(32, 13, 2023)
print("Date invalide est valide?", date_invalide.est_valide())
```
