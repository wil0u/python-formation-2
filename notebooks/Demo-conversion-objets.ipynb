Comment convertir des objets de base

```python
x = 1
# 1
x = str(x)
# '1'
x = int(x)
# 1
x = float(x)
# 1.0
x = bool(x)
# True
x = int(x)
# 1
```

```python
# ne marche pas
y = 'True'
y = int(y)
```

```python
# marche
y = bool(y)
y = int(y)
```
