```python
# query select , attention, ne pas lancer un select * cela peut dépasser la mémoire de python
sql1 = "select * from customers2 LIMIT 10;"

# query count
sql2 = "SELECT count(*) FROM customers2;"

# rapatrier les données en local
dat = sqlio.read_sql_query(sql1, conn)
nblignes = sqlio.read_sql_query(sql2, conn)
conn = None
```
