```python
# Vérifions que tout est objet
examples = [
    42,           # entier
    3.14,         # flottant
    "hello",      # chaîne de caractères
    [1, 2, 3],    # liste
    (1, 2, 3),    # tuple
    {'a': 1},     # dictionnaire
    {1, 2, 3},    # ensemble
    lambda x: x,  # fonction
    True          # booléen
]

print("Démonstration que tout est objet en Python:")
for obj in examples:
    print(f"{str(type(obj)):20} | id: {id(obj):20}")
```
