# OBJET EN MEMOIRE ET OCCUPATION DE LA MEMOIRE

```python
# deux variables différentes qui pointent sur la meme donnée
var1 = 5
var2 = 5
var3 = 5.0
var4 =var1
```

```python
# Affiche l'adresse mémoire de localisation de la donnée
# python variables sont des pointeurs
print(hex(id(var1)))
print(hex(id(var2)))
print(hex(id(var3)))
print(hex(id(var4)))
```

c'est la meme adresse pour les mêmes valeurs

```python
# si on modifie la valeur, l'emplacement change
```

```python
var1 = var1**2
```

```python
print(hex(id(var1)))
print(hex(id(var2)))
print(hex(id(var3)))
print(hex(id(var4)))
```

```python
print(var1,var2,var3,var4)
```
