Une longue instruction if

```python
my_first_variable = 10
my_second_variable = 20
my_third_and_last_variable = 30
```

Utilisation du caractère de continuation de ligne \

```python
if my_first_variable > 5 and \
   my_second_variable > 15 and \
   my_third_and_last_variable > 25 or \
   my_first_variable < my_second_variable and \
   my_first_variable < my_third_and_last_variable and \
   my_third_and_last_variable > 25 :
    print("Toutes les conditions sont remplies.")
```

Autre manière sans utiliser le caractère de continuation de ligne<br>
Les parenthèses permettent à Python de savoir que l'instruction se poursuit sur la ligne suivante

```python
if (my_first_variable > 5 and 
   my_second_variable > 15 and 
   my_third_and_last_variable > 25 or 
   my_first_variable < my_second_variable and 
   my_first_variable < my_third_and_last_variable and 
   my_third_and_last_variable > 25) :
    print("Toutes les conditions sont remplies.")
```

```python
for i in range(1, 11):
    if i % 2 == 0: 
        print("Le nombre", i, "est pair")
        if i % 4 == 0:
            print("De plus, le nombre", i, "est divisible par 4")
            print("Hello")
        else:
            print("Le nombre", i, "est impair")
            print("Hey !")
```
