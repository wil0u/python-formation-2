## 1er cas : Classe carré
Créer une classe MaClasseCarre qui contient les caractéristiques suivants :
 - 1 paramètre : nombre initial
 - 2 attributs :
  - param_nombre : la valeur du paramètre
  - attribut_carre : la valeur du paramètre au carré si calculé, -1 sinon
 - 2 méthodes :
  - carre : met le paramètre initial au carré
  - une surcharge de \__str__ : "nombre en entrée : ..., carré en sortie : ..."

```python
# une classe a un constructeur et 1 à n méthodes
class MaClasseCarre():
    # constructeur de la classe : ce qui est fabriqué quand on appelle la classe, cela construit une instance
    def __init__(self, param_nombre):
        self.param_nombre = param_nombre
        self.attribut_carre = -1               
    
    # fonction principale 
    def carre(self):
        self.attribut_carre = self.param_nombre**2
        return self
    
    # controle ce qu'on veut afficher si on print la classe 
    def __str__(self):
        return f"nombre en entrée : {self.param_nombre}, carré en sortie : {self.attribut_carre}"
```

```python
# quand on lance la classe, cela exécute la méthodec _init_ 
mon_objet = MaClasseCarre(5) # affiche l'empl mem
print(mon_objet)
# donc en sortie ici, on a pas appelé la méthode carré donc le carré n'est pas calculé
```

```python
# Accès aux attributs
print(mon_objet.param_nombre)
print(mon_objet.attribut_carre)
```

```python
# la méthode déclenche bien le calcul
mon_objet.carre()
print(mon_objet)
```

## 2ème cas : classe produit vectorielle
Créer une classe MaCombinaisonLineaire qui contient les caractéristiques suivantes :
 - 2 paramètres :
  - 1 vecteur d'individu (liste de chiffres)
  - 1 vecteur de coefficient (liste de chiffres)
 - 3 attributs :
  - 1 pour chaque paramètre
  - 1 pour le résultat de la combinaison linéraire, -1 sinon
 - 2 méthodes :
  - comb_lin : fonction qui prend les 2 listes et faire le produit scalaire (np.dot)
  - une surcharge de \__str__ : "nombre de coeffs : ..., liste des individus : ..., résultat du produit cartésien : ..."
  
---
**NOTE**
Dans une V2, gérer les cas où les listes :
 - ne sont pas de la même taille
 - ne contiennent pas que des chiffres
---

```python
class MaCombinaisonLineaire():
    def __init__(self, list_X,list_Coef):
        self.list_X = list_X
        self.list_Coef = list_Coef
        # on declare lattribut prediction
        self.prediction = -1
    
    # fonction principale, combinaison lin
    def comb_lin(self):
        try:
            self.prediction = np.dot(self.list_X,self.list_Coef)
            return self # on retourne tout, si on fait return prediction ca ne marche pas
        except ValueError:            
            self.prediction=float('nan') # on met vide
            return self # on retourne les resultats quand lexcept est raise pour pouvoir afficher le print (__str__)

    
    # controle ce qu'on veut afficher si on print la classe 
    def __str__(self):
        return f"nombre de coeff : {len(self.list_Coef)}, liste des individus: {self.list_X}, résultat du produit cartésien: {self.prediction}"
```

```python
# Test 1 :
ind = [1, 0, 0, 1]
coef = [0.25, 0.25, 0.25, 0.25]
mon_objet = MaCombinaisonLineaire(ind,coef)
print(mon_objet)
mon_objet.comb_lin()
print(mon_objet)
```

```python
# Test 2 :
ind = [1, 0, 0, 1]
coef = [0.25, 0.25, 0.25]
mon_objet = MaCombinaisonLineaire(ind,coef)
print(mon_objet)
mon_objet.comb_lin()
print(mon_objet)
```
