Accéder aux arguments par position:

```python
print('{0}, {1}, {2}'.format('a', 'b', 'c'))  # affiche : 'a, b, c'
print('{}, {}, {}'.format('a', 'b', 'c'))  # 2.7+ uniquement, affiche : 'a, b, c'
print('{2}, {1}, {0}'.format('a', 'b', 'c'))  # affiche : 'c, b, a'
print('{2}, {1}, {0}'.format(*'abc'))  # déballage de la séquence d'arguments, affiche : 'c, b, a'
print('{0}{1}{0}'.format('abra', 'cad'))   # les indices des arguments peuvent être répétés, affiche : 'abracadabra'
```

Accéder aux arguments par nom :

```python
print('Coordinates: {latitude}, {longitude}'.format(latitude='37.24N', longitude='-115.81W'))  # affiche : 'Coordinates: 37.24N, -115.81W'
coord = {'latitude': '37.24N', 'longitude': '-115.81W'}
print('Coordinates: {latitude}, {longitude}'.format(**coord))  # affiche : 'Coordinates: 37.24N, -115.81W'
```

Accéder aux attributs des arguments :

```python
c = 3-5j
print('Le nombre complexe {0} est formé à partir de la partie réelle {0.real} et de la partie imaginaire {0.imag}.'.format(c))  # affiche : 'Le nombre complexe (3-5j) est formé à partir de la partie réelle 3.0 et de la partie imaginaire -5.0.'
```

Accéder aux éléments des arguments :

```python
coord = (3, 5)
print('X: {0[0]};  Y: {0[1]}'.format(coord))  # affiche : 'X: 3;  Y: 5'
```

Remplacement de %s et %r :

```python
print("repr() affiche des guillemets: {!r}; str() n'en affiche pas: {!s}".format('test1', 'test2'))  # affiche : "repr() affiche des guillemets: 'test1'; str() n'en affiche pas: test2"
```

Alignement du texte et spécification d'une largeur :

```python
print('{:<30}'.format('aligné à gauche'))  # affiche : 'aligné à gauche                  '
print('{:>30}'.format('aligné à droite'))  # affiche : '                 aligné à droite'
print('{:^30}'.format('centré'))  # affiche : '           centré            '
print('{:*^30}'.format('centré'))  # utilise '*' comme caractère de remplissage, affiche : '***********centré***********'
```

Remplacement de %+f, %-f, et % f et spécification d'un signe :

```python
print('{:+f}; {:+f}'.format(3.14, -3.14))  # toujours afficher le signe, affiche : '+3.140000; -3.140000'
print('{: f}; {: f}'.format(3.14, -3.14))  # afficher un espace pour les nombres positifs, affiche : ' 3.140000; -3.140000'
print('{:-f}; {:-f}'.format(3.14, -3.14))  # afficher uniquement le signe négatif -- identique à '{:f}; {:f}', affiche : '3.140000; -3.140000'
```

Remplacement de %x et %o et conversion de la valeur en différentes bases :

```python
print("int: {0:d};  hex: {0:x};  oct: {0:o};  bin: {0:b}".format(42))  # affiche : 'int: 42;  hex: 2a;  oct: 52;  bin: 101010'
print("int: {0:d};  hex: {0:#x};  oct: {0:#o};  bin: {0:#b}".format(42))  # avec 0x, 0o, ou 0b comme préfixe, affiche : 'int: 42;  hex: 0x2a;  oct: 0o52;  bin: 0b101010'
```

Utilisation de la virgule comme séparateur de milliers :

```python
print('{:,}'.format(1234567890))  # affiche : '1,234,567,890'
```

Expression d'un pourcentage :

```python
points = 19.5
total = 22
print('Réponses correctes: {:.2%}'.format(points/total))  # affiche : 'Réponses correctes: 88.64%'
```

Utilisation d'un formatage spécifique au type :

```python
import datetime
d = datetime.datetime(2010, 7, 4, 12, 15, 58)
print('{:%Y-%m-%d %H:%M:%S}'.format(d))  # affiche : '2010-07-04 12:15:58'
```

Imbrication d'arguments et exemples plus complexes :

```python
for align, text in zip('<^>', ['gauche', 'centre', 'droite']):
    print('{0:{fill}{align}16}'.format(text, fill=align, align=align))  # affiche : 'gauche<<<<<<<<<<', '^^^^^centre^^^^^', '>>>>>>>>>>droite'
```

```python
octets = [192, 168, 0, 1]
print('{:02X}{:02X}{:02X}{:02X}'.format(*octets))  # affiche : 'C0A80001'
print(int('C0A80001', 16))  # affiche : 3232235521
```

```python
largeur = 5
for num in range(5,12):
    for base in 'dXob':
        print('{0:{width}{base}}'.format(num, base=base, width=largeur), end=" ")
    print()
```
