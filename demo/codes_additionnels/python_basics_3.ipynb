c = a + b

```python
if (c := a + b) > 10:
    print(c)  # Affiche 13 
```

ne liste arbitrairement grande de nombre

```python
my_list = [1,5,23,5639,232,53938,23,54,10,239737,5,22222,44343431,534,212]
```

Test si à l'intérieur de ma liste il y a au moins 3 nombres impaires <br>
et que la somme de ces nombre est supérieur à 1000 et que la moyenne est de <br>
ces nombre est positive. 

```python
if (taille :=len(nombre_impair := [x for x in my_list if x%2 != 0] ) ) >=3 and \
      (somme := sum(nombre_impair)) > 1000 and \
          (moy:= somme/taille) > 0:
    
    print(my_list)
    print(nombre_impair) 
    print(taille)
    print(somme)
    print(moy)
```

If ternaire

```python
    
#Une liste arbitrairement grande de nombre
my_list = [1,5,23,5639,232,53938,23,54,10,239737,5,22222,44343431,534,212]
```

Test si à l'intérieur de ma liste il y a au moins 3 nombres impaires <br>
et que la somme de ces nombre est supérieur à 1000 et que la moyenne est de <br>
ces nombre est positive. 

```python
condition = (taille :=len(nombre_impair := [x for x in my_list if x%2 != 0] ) ) >=3 and \
      (somme := sum(nombre_impair)) > 1000 and \
          (moy:= somme/taille) > 0
```

```python
resultat = "la condition est vrai" if condition else "la condition est fausse"
```

Une fonction qui en 1 instruction  : Fait 3 tests booléens, 4 affectations, retourne ces 4 variables si les 3 tests sont OK et retourne None sinon.

```python
my_func = lambda l : (nombre_impair,taille,somme,moy) if \
    (taille :=len(nombre_impair := [x for x in l if x%2 != 0] ) ) >=3 and \
      (somme := sum(nombre_impair)) > 10 and \
          (moy:= somme/taille) > 0 else \
            None
```
