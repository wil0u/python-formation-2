# classe qui à partir d'un nombre renvie son carré

```python
# une classe a un constructeur et 1/n méthodes
class mafonction_carre():
    # constructeur de la classe : ce qui est fait quand on appelle la classe
    def __init__(self, param_nombre):
        self.param_nombre = param_nombre
        self.attribut_carre = -1               
    
    # fonction principale, met à jour le param_nombre
    def carre(self):
        self.attribut_carre = self.param_nombre**2
        return self
    
    # controle ce qu'on veut afficher si on print la classe 
    def __str__(self):
        return f"nombre en entrée : {self.param_nombre}, carré en sortie: {self.attribut_carre}"
```

## appel de la classe

```python
# quand on lance la classe, cela exécute la méthodec _init_ 
mafonction_carre(5) # affiche l'empl mem
print(mafonction_carre(5))
# donc en sortie ici, on a pas appelé la méthode carré donc le carré n'est pas calculé
```

## appel de la méthode

```python
# la méthode déclenche bien le calcul
print(mafonction_carre(5).carre())
```

```python
# classe qui à partir d'un vecteur d'un individu et d'un vecteur de coeff calcule la comb lin
# ajout test que le nb delements des deux listes doit etre egal
import numpy as np
class mafonction_predictlin():
    def __init__(self, list_X,list_Coef):
        self.list_X = list_X
        self.list_Coef = list_Coef        
        # on declare lattribut prediction
        self.prediction = -1
        self.err=-1
    
    # fonction principale, combinaison lin
    def comb_lin(self):
        try:            
            self.prediction = np.dot(self.list_X,self.list_Coef)
            return self #quand on appell on retourne tout
        except ValueError:            
            print("les deux listes n'ont pas la meme longueur")
            self.err=999 # on def un code err
            self.prediction=float('nan') # on met vide
            return self # on retourne les resultats quand lexcept est raise pour pouvoir afficher le print (__str__)
    
    # controle ce qu'on veut afficher si on print la classe 
    def __str__(self):
        return f"nb de variables : {len(self.list_Coef)}, Ind à prédire: {self.list_X}, Comb lin. en sortie: {self.prediction}, Code Erreur: {self.err}"
```

```python
# deux longueurs diff
ind = [1, 0,0,1,0,1]
coef = [0.25, 0.25,0.25, 0.25]
print( mafonction_predictlin(ind,coef).comb_lin())
```
