## STEP 3 : Utiliser les regex pour corriger toutes les anomalies identifiées

```python
data_output = rep+"/out.txt"
```

```python
# ouvrir le fichier data_output en mode écriture
fout = open(data_output, "wt")
```

```python
# Parcourir chaque ligne pour lui appliquer l'ensemble des correctifs (en partant de l)
nom_columns = []
nb_columns = 0
for index, text in enumerate(l):
    if index == 0 : # si première ligne, on identifie le nombre de colonnes
        nom_columns = text.split(";")
        nb_columns = len(nom_columns)
    text_propre = text.replace('\t','') \
                        .replace('\n','') \
                        .replace('"', '') \
                        .replace('\ufeff','')
    text_propre_split = text_propre.split(";")
    while len(text_propre_split)<3:
        text_propre_split.append("")
    text_propre_final = ";".join(text_propre_split[0:3])+'\n'
    #ecrit la ligne propre
    fout.write(text_propre_final)
```

```python
# On ferme les 2 fichiers ouverts
f.close()
fout.close()
```
