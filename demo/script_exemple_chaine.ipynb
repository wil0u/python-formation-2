Définir une chaîne de caractères

```python
s = "Hello, World! Hello, Python!"
```

S.count(sub[, start[, end]])

```python
sub = "Hello"
print(f"Count of '{sub}' in '{s}' : ", s.count(sub))
```

S.find(sub [,start [,end]])

```python
sub = "World"
print(f"Index of '{sub}' in '{s}' : ", s.find(sub))
```

S.index(sub [,start [,end]])

```python
sub = "Python"
print(f"Index of '{sub}' in '{s}' : ", s.index(sub))
```

S.lower()

```python
print("Lowercase : ", s.lower())
```

S.replace(old, new[, count])

```python
old = "Hello"
new = "Hi"
print(f"Replace '{old}' with '{new}' in '{s}' : ", s.replace(old, new))
```

S.split([sep [,maxsplit]])

```python
sep = " "
print(f"Split '{s}' by '{sep}' : ", s.split(sep))
```

S.strip([chars])

```python
s = "    Hello, World!   "
print(f"Strip '{s}' : '", s.strip(), "'")
```

S.upper()

```python
s = "Hello, World!"
print("Uppercase : ", s.upper())
```
