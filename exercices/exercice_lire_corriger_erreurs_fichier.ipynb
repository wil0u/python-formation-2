# Il  y a toujours le retour chariot
on va lire tout le contenu, pas par ligne et on cible le retour chariot avec une regex

```python
import re
#lire le contenu sans ligne à ligne et supprimer les retours chariots
fin = open(rep+"\\exercice_RH\\JOBS_jouet_aparse.csv", "rt")
fout = open(rep+"\\out.txt", "wt")
```

```python
content = fin.read()

# la regex : 
content=re.sub(';"\n',';"',content).replace('\t','').replace('ï»¿','').replace('"', '')

content[0:2000]
```

```python
#ecrit dans le fichier sortie
fout.write(content)
#close input and output files
fin.close()
fout.close()
```
