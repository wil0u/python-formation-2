# LAMBDA FUNCTION
Elles s'utilisent à la volée dans une fonction map ; apply, applymap, filter qui s'applique la fonction définie dans le lambda à chaque élément d'un objet iterable

```python
# on crée un tuple
numbers = (1, 2, 3, 4)
```

```python
# utilise la lambda func pour à tout x renvoyer f(x)=x²
# map applique une fonction à un iterable, ici le tuple
result = list(map(lambda x: x*x, numbers))
```

```python
# autre fonction
result= list(map(lambda x: x==1.00, numbers))
print(result)
result.count(False)
```

# Création de fonction

```python
# création d'une fonction qui renvoie le carré de x
def fonc_carre(x):
    return x**2
```

```python
fonc_carre(-10)
```

```python
# Si l'utilisateur utilise mal la fonction, on a une erreur
fonc_carre('5')
```

```python
# On va donc attraper cette erreur pour programmer un comportement face à cette erreur
def fonc_carre(x):
    try:
        return x**2
    except TypeError:
        print("x n'est pas un nombre!!!")  
    
```

```python
fonc_carre('5')
```

```python
fonc_carre(2)
```

```python
# fonction en dur avec gestion des exceptions
from math import sqrt
def _sqrt(x):
    try:
        x>=0
        return sqrt(x)
    except:
        x<0
        print("calcul impossible")
    
    # test
    _sqrt(-5)
    _sqrt(+5)
```

```python
result=list(map(_sqrt, numbers))  # Output [2, 4, 6, 8]
print(result)
```

# Exemple fonction standardisation

```python
# créons un vecteur de données numérique en objet array
import numpy as np
Matrice = np.array(range(10))
```

```python
# créons une fonction qui standardise un vecteur (type 1darray) : centrer et réduire (soustraire la moyenne et diviser par l'écart type)
# x doit etreu un np.array car on applique dessus np.mean ...
def standardisation(x): 
        x-=np.mean(x) # the -= means can be read as x = x- np.mean(x)
        x/=np.std(x)
        return x
```

```python
# si on applique la fonction sur un array non float il y a un erreur
standardisation(Matrice)
```

```python
# appliquons la fonction standardisation, attention aux types des données
standardisation(Matrice.astype(float))
```

```python
# Introduire une gestion d'exception dans la fonction => try except
def standardisation(x): 
    try:        
        x-=np.mean(x) # the -= means can be read as x = x- np.mean(x)
        x/=np.std(x)
        return x
    except TypeError:
        print("le type des données n'est pas float")
```

```python
standardisation(Matrice)
```

```python
standardisation(Matrice.astype(float))
```

```python
# si matrice2 n'est pas un array mais un tuple
# cela marche car np.mean fonctionne sur une liste de val numériques
matrice2=(1,2,3,4)
standardisation(matrice2)
```
